```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print("NaNs in TotalCharges after conversion:", df['TotalCharges'].isnull().sum())
 
print(df['Churn'].value_counts())
print(df['Churn'].value_counts(normalize=True))
 
print("Duplicates:", df.duplicated().sum())
 
print(df.isnull().sum()[df.isnull().sum() > 0])
 
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()
print("Numeric:", num_cols)
print("Categorical:", cat_cols)
 
print("Unique customerIDs:", df['customerID'].nunique(), "out of", len(df))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 

num_features = ['tenure', 'MonthlyCharges', 'TotalCharges']
fig, axes = plt.subplots(1, 3, figsize=(18, 4))
for ax, col in zip(axes, num_features):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(f"Distribution of {col}")
plt.show()
```

```python
df[num_features].skew()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 4))
for ax, col in zip(axes, num_features):
    sns.boxplot(x=df[col], ax=ax)
    ax.set_title(f'Boxplot of {col}')
plt.tight_layout()
plt.show()
```

```python
for col in cat_cols:
    if col in ['CustomerID0', 'Churn']:
        continue
    print(f'---{col}')
    print(df[col].value_counts(dropna=False))
    print()
```

```python
cat_features = [c for c in cat_cols if c not in ['customerID', 'Churn']]

fig, axes = plt.subplots(4, 4, figsize=(20, 16))
axes = axes.flatten()
for ax, col in zip(axes, cat_features):
    sns.countplot(data=df, x=col, ax=ax)
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)
for ax in axes[len(cat_features):]:
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
numeric_features = ['tenure', 'MonthlyCharges', 'TotalCharges']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, col in zip(axes, numeric_features):
    sns.boxplot(data=df, x='Churn', y=col, ax=ax)
    ax.set_title(f'{col} vs Churn')
plt.tight_layout()
plt.show()
```

```python
df.groupby('Churn')[numeric_features].mean()
```

```python
cat_features = [c for c in cat_cols if c not in ['customerID', 'Churn']]

for col in cat_features:
    churn_rate = df.groupby(col)['Churn'].apply(
        lambda x: (x == 'Yes').mean()
    ).sort_values(ascending=False)
    print(f"--- {col} ---")
    print(churn_rate)
    print()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, col in zip(axes, ['Contract', 'InternetService', 'PaymentMethod']):
    sns.countplot(data=df, x=col, hue='Churn', ax=ax)
    ax.set_title(f'{col} vs Churn')
    ax.tick_params(axis='x', rotation=30)
plt.show()
```
